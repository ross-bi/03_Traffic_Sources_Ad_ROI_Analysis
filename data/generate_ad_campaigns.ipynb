```python
import pandas as pd
import numpy as np

np.random.seed(42)

# 定義各 channel 的 campaign 清單
campaigns_map = {
    'google':   ['Google_Brand_Search', 'Google_Shopping', 'Google_Display'],
    'facebook': ['FB_Retargeting', 'FB_Lookalike', 'FB_Awareness'],
    'email':    ['Email_Newsletter', 'Email_Promo', 'Email_Winback'],
    'organic':  ['Organic_SEO', 'Organic_Social']
}

rows = []
for channel, camp_list in campaigns_map.items():
    for camp in camp_list:
        impressions = np.random.randint(10000, 200000)
        clicks      = np.random.randint(500, min(8000, impressions))
        ad_spend    = round(np.random.uniform(500, 5000), 2)
        rows.append({
            'campaign_name': camp,
            'channel':       channel,
            'ad_spend':      ad_spend,
            'impressions':   impressions,
            'clicks':        clicks,
        })

df = pd.DataFrame(rows)

# 儲存到 data/ 資料夾（Notebook 本身已在 data/ 裡）
df.to_csv('ad_campaigns.csv', index=False)

print(f"✅ 已生成 {len(df)} 筆資料")
print(df)
```
